```python
# importing Libraries
import pandas as pd 
from datasets import load_dataset
import matplotlib.pyplot as plt 

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()

# Data Cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

# pandas.DataFrame.fillna
### DataFrame.fillna(value, *, axis=None, inplace=False, limit=None)

#### fillna() → Fill missing values

```python
df.loc[:10, 'salary_year_avg':'salary_hour_avg']
```

```python
median_salary_year = df['salary_year_avg'].median()
```

```python
median_salary_hour = df['salary_hour_avg'].median()
```

```python
df_filled = df

df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour).round(2)
```

```python
df_filled.loc[:10, 'salary_year_avg':'salary_hour_avg']
```

# pandas.DataFrame.drop_duplicates
### DataFrame.drop_duplicates(subset=None, *, keep='first', inplace=False, ignore_index=False)

```python
df_unique = df_filled

df_unique = df_unique.drop_duplicates()

print('Length of orginal df:           ', len(df_filled))
print('Length of drop duplicateds df:', len(df_unique))
print('Row Dropped:                 ', len(df_filled)-len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])
print('Length of orginal df:           ', len(df_filled))
print('Length of drop duplicateds df:', len(df_unique))
print('Row Dropped:                 ', len(df_filled)-len(df_unique))
```

## Remove Rows with Missing Values

#### dropna() → Remove missing values

```python
rows_before = len(df)
df_cleaned = df.dropna(subset= ['salary_year_avg'])
rows_after= len(df_cleaned)

rows_before, rows_after
```
