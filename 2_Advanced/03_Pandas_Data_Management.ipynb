### Drop Missing Values

```python
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df_copy = df.copy()
df_cleaned = df_copy.dropna(subset=['salary_year_avg'])
df_cleaned.head(5)
```
